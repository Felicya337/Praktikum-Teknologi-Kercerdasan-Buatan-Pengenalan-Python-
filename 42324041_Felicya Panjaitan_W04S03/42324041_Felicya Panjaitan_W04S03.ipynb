```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
iris = load_iris()
```

Kode ini digunakan untuk mengimpor library yang diperlukan dan memuat dataset Iris dari Scikit-learn untuk digunakan dalam proses klasifikasi.

```python
df = pd.DataFrame(iris.data, columns=iris.feature_names)
df["Species"] = iris.target_names[iris.target]
X = df.drop(columns=["Species"])
y = df["Species"]
print(X)
```

Kode ini digunakan untuk **mengubah dataset Iris menjadi DataFrame, menambahkan label species, serta memisahkan data menjadi fitur (X) dan target (y) untuk proses klasifikasi.

```python
y
```

Output ini menunjukkan label species pada dataset Iris, yaitu setosa, versicolor, dan virginica, dengan total 150 data.

```python
from sklearn.preprocessing import LabelEncoder
lab = LabelEncoder()
y = lab.fit_transform(y)
y
```

Kode ini digunakan untuk mengubah label species dari bentuk teks menjadi angka, yaitu setosa = 0, versicolor = 1, dan virginica = 2, agar dapat diproses oleh model.

```python
from sklearn.model_selection import train_test_split
X_train, X_test = train_test_split(X, test_size=0.2, random_state=0)
y_train, y_test = train_test_split(y, test_size=0.2, random_state=0)
```

Kode ini digunakan untuk membagi dataset menjadi data training sebesar 80% dan data testing sebesar 20% untuk melatih dan menguji model.

```python
from sklearn.naive_bayes import GaussianNB
model = GaussianNB()
NB_model = model.fit(X_train, y_train)
```

Kode ini digunakan untuk membuat model Gaussian Naïve Bayes dan melatihnya menggunakan data training (X_train dan y_train).

```python
result = NB_model.predict(X_test)
result
```

Kode ini digunakan untuk memprediksi species pada data testing menggunakan model Gaussian Naïve Bayes yang sudah dilatih.

```python
from sklearn.metrics import accuracy_score
print('Accuracy of data testing: ', accuracy_score(y_test, result)*100)
```

Kode ini digunakan untuk menghitung akurasi model pada data testing, dengan hasil akurasi sebesar 96,67%.

```python
input_data = [[5.1, 3.5, 1.4, 0.2]]
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Kode ini digunakan untuk memprediksi species dari data baru, dengan hasil prediksi yaitu setosa. Warning muncul karena input tidak memiliki nama fitur yang sama dengan data training.

```python
input_data = pd.DataFrame(np.array([[5.1, 3.5, 1.4, 0.2]]), columns=iris.feature_names)
result = NB_model.predict(input_data)
prediction_label = lab.inverse_transform(result)
print('Hasil prediksi untuk input baru adalah:', prediction_label)
```

Kode ini digunakan untuk memprediksi species data baru dengan format fitur yang sesuai dengan model, sehingga hasil prediksinya adalah setosa tanpa warning.
